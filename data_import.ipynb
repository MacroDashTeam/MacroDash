```python
def getFRED(code: str, firstDay: pd.Timestamp) -> pd.Series:
    data = fred.get_series(code).dropna()
    data = data.loc[data.index > firstDay]
    return data

def getYF(code: str, firstDay: pd.Timestamp) -> pd.Series:
    data = yf.Ticker(code).history("5y")["Close"].dropna()
    data = data.loc[data.index.tz_convert(None) > firstDay.to_numpy()]
    return data

getSeries = {
    "FRED":          getFRED,
    "Yahoo Finance": getYF}

def makeTable(codes: list[str], horizon: pd.DateOffset) -> pd.DataFrame:
    numberOfSeries = len(codes)
    table = pd.DataFrame(data = getSeries[series.loc[series["code"] == codes[0]]["source"].iat[0]](codes[0], firstDay = pd.Timestamp.now() - horizon))
    table.columns = [codes[0]]
    seriesNumber = 1
    while seriesNumber < numberOfSeries:
        other = getSeries[series.loc[series["code"] == codes[seriesNumber]]["source"].iat[0]](codes[seriesNumber], firstDay = pd.Timestamp.now() - horizon)
        other.name = codes[seriesNumber]
        table = table.join(other = other, how = "outer")
        seriesNumber += 1
    return table
```

```python
for codes in [["DFF", "DFEDTARU", "DFEDTARL", "SOFR", "IORB"], # left column
              ["MORTGAGE30US"],
              ["DGS10", "T10Y2Y"],
              ["M2SL"],
              ["UNRATE"],
              ["^GSPC"], # right column
              ["^VIX"],
              ["GC=F"],
              ["XLF", "XLY", "XLC", "XLV", "XLI", "XLP", "XLE", "XLB", "XLRE", "XLU"]]:
    data = makeTable(codes, pd.DateOffset(years = 5))
    for code in codes: plt.plot(data.index, data[code], label = series.loc[series["code"] == code]["title"].iat[0])
    plt.legend(loc = 'upper left', bbox_to_anchor = (1, 1))
    plt.show()
```
